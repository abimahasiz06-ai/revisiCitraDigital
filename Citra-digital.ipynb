Metode: Grayscale

Bahasa: Python

```python
from PIL import Image
import matplotlib.pyplot as plt
```

```python
# 1. Membuka foto yang dibuat di Paint
foto = Image.open("/content/drive/MyDrive/pas foto.jpeg")
```

```python
# 2. Mengubah foto berwarna menjadi grayscale
foto_gray = foto.convert("L")
```

```python
# 3. Menampilkan foto asli dan grayscale
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(foto)
plt.title("Foto Asli")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(foto_gray, cmap="gray")
plt.title("Foto Grayscale")
plt.axis("off")

plt.show()
```

```python
# 4. Menyimpan hasil grayscale
foto_gray.save("foto_grayscale.png")

print("Foto berhasil diubah menjadi grayscale!")
print("Hasil disimpan sebagai foto_grayscale.png")
```
